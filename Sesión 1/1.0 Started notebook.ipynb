### Load a CSV using Spark
- This reads a CSV file into a Spark DataFrame
### Display a DataFrame visually in Databricks
- The display() function shows data in a table view


```python

df = spark.read.csv(directory_bronze, header=True, inferSchema=True) # Important header and inferSchema = TRUE

display(df)
```

### Generate basic statistics
- Use describe() to get count, mean, stddev, min, max for numeric columns

```python
display(df.describe())
```

```python
# Show the schema (column names and types)
df.printSchema()
```

```python
# Show number of rows and columns
row_count = df.count()
col_count = len(df.columns)
print(f"This dataset contains {row_count} rows and {col_count} columns.")
```

```python
# Check for nulls in each column
print("Null values per column:")
from pyspark.sql.functions import col, sum as spark_sum
null_counts = df.select([spark_sum(col(c).isNull().cast("int")).alias(c) for c in df.columns])
display(null_counts)
```

Some transformations using pyspark and creating new data. Check how import works

```python
from pyspark.sql.functions import col, round

# Add a new column: Revenue per Salary ratio
emp_df = df.withColumn("RevenuePerSalary", round(col("MonthlyRevenue_kEUR") / col("Salary_kEUR"), 2))

display(emp_df)
```

```python
# Filter: Only keep rows where RevenuePerSalary > 2
filtered_df = emp_df.filter(col("RevenuePerSalary") > 2)

display(filtered_df)
```

```python
# Sort by RevenuePerSalary descending
sorted_df = filtered_df.orderBy(col("RevenuePerSalary").desc())

# Show result
display(sorted_df)
```

### Save DataFrame to a new CSV file
Could be used other format distint to csv

```python
# Define output table name for saving processed data
table_name = "workspace.default.employee_stats_silver"
```

```python
# Save as a Delta table in Unity Catalog (workspace catalog)
# This is the recommended approach for Serverless compute
table_name = "workspace.default.employee_stats_silver"
sorted_df.write.mode("overwrite").saveAsTable(table_name)

print(f"✓ Data saved as table: {table_name}")
print(f"  Rows written: {sorted_df.count()}")
display(df)
```

```python
# Query the saved Delta table to verify the data
print(f"Querying table: {table_name}")
display(spark.table(table_name))
```
